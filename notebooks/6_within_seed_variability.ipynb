```python
import os, sys, pickle
import matplotlib.pyplot as plt
import numpy as np
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))

project_root = os.path.abspath(os.path.join(os.getcwd(), '../'))
```

## Expert

```python
expert_data_path = os.path.join(project_root, 'results', 'expert', 'within_seed_var')
expert_data_files = os.listdir(expert_data_path)
expert_data = {}

for file in expert_data_files:
    file_path = os.path.join(expert_data_path, file)
    with open(file_path, 'rb') as f:
        R_s_r = pickle.load(f)

    for s_r, val in R_s_r.items():
        if s_r[0] not in expert_data:
            expert_data[s_r[0]] = []

        # within seeds accumulated rewards
        expert_data[s_r[0]].append(np.sum(val))
```

```python
for env_seed, acc_R in expert_data.items():
    expert_data[env_seed] = (np.mean(acc_R), np.std(acc_R))
    if expert_data[env_seed][1] > 10:
        print(env_seed)
```

```python
# unpack data
seeds = list(expert_data.keys())
means = [expert_data[s][0] for s in seeds]
stds  = [expert_data[s][1] for s in seeds]

mean_of_stds = np.round(np.mean(stds), 2)
std_of_stds = np.round(np.std(stds), 2)
mean_of_means = np.round(np.mean(means), 2)

# plot
plt.figure()
plt.bar(seeds, means, yerr=stds, capsize=5)

plt.xlabel("environment seed")
plt.ylabel("accumulated reward")
plt.title(f"mean of means {mean_of_means} / mean of stds {mean_of_stds} / std of stds {std_of_stds}")
plt.ylim(0, 350)

plt.tight_layout()
plt.show()
```
